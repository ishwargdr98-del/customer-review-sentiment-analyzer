```python
reviews = [
    "This product is very good",
    "I love this product",
    "Excellent quality",
    "This product is bad",
    "I hate this product",
    "Very poor quality"
]

sentiments = [
    "positive",
    "positive",
    "positive",
    "negative",
    "negative",
    "negative"
]

print(reviews)
print(sentiments)
```

```python
import pandas as pd

df=pd.DataFrame({"reviews":reviews,"sentiments":sentiments})

print(df)
print(df.shape)
```

```python

print("Dataset information:")
print("Rows and columns:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nSentiment counts:")
print(df["sentiments"].value_counts())
```

```python
from numpy import vectorize
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer=TfidfVectorizer()

x=vectorizer.fit_transform(df["reviews"])

print("Feature matrix shape:",x.shape)
print("Words learned:",vectorizer.get_feature_names_out())
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

y=df["sentiments"]

print(y)
```

```python
model.fit(x,y)

print("Model training complete!")
```

```python
new_review = ["This product is long"]

review_features = vectorizer.transform(new_review)

predict = model.predict(review_features)


print("Review:", new_review[0])
print("Predicted sentiment:", predict[0])


```

```python
print("Words:",vectorizer.get_feature_names_out())
print("Learned weights:",model.coef_)
print("Class order:",model.classes_)
```

```python
for word,weight in zip(vectorizer.get_feature_names_out(),model.coef_[0]):
  print(f"{word}:{weight:.3f}")
```

```python
def predict_sentiment(review):
  review_features = vectorizer.transform([review])
  predict = model.predict(review_features)
  return predict[0]

print(predict_sentiment("This product is excellent"))
print(predict_sentiment("This product is poor"))
```

```python
review =input('Apna product review likho: ')

result = predict_sentiment(review)

print(f"Review : {review}")
print(f"Predicted Sentiment:, {result}")
```
